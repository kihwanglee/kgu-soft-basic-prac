## 1. 딕셔너리 만들고 꺼내기

**딕셔너리(dictionary)** 는 **이름(키, key)** 으로 값을 찾는 자료형입니다. 리스트가 `[번호]` 로 값을 찾았다면, 딕셔너리는 `["키"]` 로 찾습니다.

- `{키: 값, 키: 값, ...}` 형태로 중괄호 `{}` 안에 **키와 값의 짝**을 쉼표로 나열합니다.
- 전화번호부에서 **이름으로 번호를 찾는 것**과 같습니다.

**딕셔너리 만들기** — 과목명(키)에 점수(값)를 짝지어 저장합니다.

````python
scores = {"국어": 85, "수학": 92, "영어": 78}
print(scores)
print(type(scores))   # <class 'dict'>
````

**값 꺼내기** — `딕셔너리["키"]` 로 그 키에 짝지어진 값을 꺼냅니다.

> ⚠️ **주의**: 없는 키(`scores["과학"]`)를 꺼내려 하면 `KeyError` 가 납니다.

````python
scores = {"국어": 85, "수학": 92, "영어": 78}
print(scores["국어"])   # 85
print(scores["영어"])   # 78
````

## 2. 안전하게 꺼내기 `get()` 과 추가·수정·삭제

- `딕셔너리.get("키")` 는 키가 없어도 **오류 없이** `None` 을 돌려줍니다.
- `딕셔너리.get("키", 기본값)` 으로 없을 때 돌려줄 **기본값**을 정할 수 있습니다.

````python
scores = {"국어": 85, "수학": 92, "영어": 78}
print(scores.get("국어"))      # 85
print(scores.get("과학"))      # None  (오류 아님)
print(scores.get("과학", 0))   # 0     (없으면 기본값 0)
````

**값 추가·수정·삭제**

- `딕셔너리["키"] = 값` → 새 키 **추가** 또는 기존 값 **수정**
- `del 딕셔너리["키"]` → 키-값 짝 **삭제**

````python
scores = {"국어": 85, "수학": 92, "영어": 78}
scores["수학"] = 100    # 기존 값 수정
scores["과학"] = 88     # 새 키 추가
del scores["영어"]      # 키-값 삭제
print(scores)
````

## 3. 키 확인·개수와 `keys()`·`values()`·`items()`

- `"키" in 딕셔너리` → 키가 있는지 확인 (`True`/`False`)
- `len(딕셔너리)` → 키-값 짝의 개수
- `keys()`·`values()`·`items()` → 각각 모든 **키**·**값**·**(키, 값) 짝**

````python
scores = {"국어": 85, "수학": 92, "영어": 78}
print("수학" in scores)      # True
print("과학" in scores)      # False
print(len(scores))           # 3
````

````python
scores = {"국어": 85, "수학": 92, "영어": 78}
print(scores.keys())     # dict_keys(['국어', '수학', '영어'])
print(scores.values())   # dict_values([85, 92, 78])
print(scores.items())    # dict_items([('국어', 85), ...])
````

## 4. `for` 문으로 딕셔너리 순회

- `for 키 in 딕셔너리:` → **키**를 하나씩 꺼냅니다. 키로 값을 찾아 함께 씁니다.
- `for 키, 값 in 딕셔너리.items():` → **키와 값을 한꺼번에** 꺼냅니다 (더 간결).

````python
scores = {"국어": 85, "수학": 92, "영어": 78}
for subject in scores:
    print(f"{subject}: {scores[subject]}점")
````

````python
scores = {"국어": 85, "수학": 92, "영어": 78}
for subject, score in scores.items():
    print(f"{subject}: {score}점")
````

## 5. `for`·`if` 로 등급 매기기와 집계

딕셔너리를 `for` 로 돌면서 `if` 로 조건을 따지면 **등급 판정·총점·평균·분류**를 할 수 있습니다.

**등급 매기기** — 점수에 따라 A/B/C/D 를 정합니다.

````python
scores = {"국어": 85, "수학": 92, "영어": 78, "과학": 65}
for subject, score in scores.items():
    if score >= 90:
        grade = "A"
    elif score >= 80:
        grade = "B"
    elif score >= 70:
        grade = "C"
    else:
        grade = "D"
    print(f"{subject}: {score}점 → {grade}")
````

**총점·평균·미달 과목** — 반복하며 총점을 누적하고, `if` 로 미달 과목을 리스트에 모읍니다.

````python
scores = {"국어": 85, "수학": 92, "영어": 78, "과학": 65, "사회": 76}
total = 0
fail_list = []
for subject, score in scores.items():
    total += score
    if score < 70:
        fail_list.append(subject)
average = total / len(scores)
print(f"총점: {total}점")
print(f"평균: {average:.1f}점")
print(f"미달 과목: {fail_list}")
````

## 6. 주문 금액 계산 (가격 딕셔너리 + 주문 리스트)

**가격 딕셔너리**와 **주문 리스트**를 함께 씁니다. `if ... in` 으로 메뉴에 있는지 확인합니다.

````python
price = {"아메리카노": 4500, "카페라테": 5000, "카푸치노": 5500, "녹차": 4000}
orders = ["아메리카노", "카페라테", "카푸치노", "아메리카노", "콜라"]

total = 0
for item in orders:
    if item in price:
        total += price[item]
        print(f"{item}: {price[item]:,}원")
    else:
        print(f"{item}: 메뉴에 없습니다.")
print(f"합계: {total:,}원")
````

## 7. 빈도 세기 패턴

**빈 딕셔너리 `{}`** 에서 시작해, 처음 보는 항목은 `1`, 이미 있으면 `+= 1` 로 셉니다. **"무엇이 몇 개"** 를 셀 때 어디서나 쓰입니다.

````python
orders = ["아메리카노", "카페라테", "카푸치노",
          "아메리카노", "녹차", "카페라테", "아메리카노"]
count = {}
for item in orders:
    if item in count:
        count[item] += 1    # 이미 있으면 1 늘리고
    else:
        count[item] = 1     # 처음 보면 1로 시작
for menu, n in count.items():
    print(f"{menu}: {n}잔")
````

**한 줄씩 따라가기** — `["봄", "여름", "봄"]` 로 `count` 가 채워지는 과정입니다.

| 반복 | `item` | `item in count`? | 반복 뒤 `count` |
|:----:|:------:|:----------------:|------------------|
| 시작 | — | — | `{}` |
| 1회 | `"봄"`   | 없음 | `{"봄": 1}` |
| 2회 | `"여름"` | 없음 | `{"봄": 1, "여름": 1}` |
| 3회 | `"봄"`   | 있음 | `{"봄": 2, "여름": 1}` |

## 8. 리스트 안의 딕셔너리

실제 데이터는 **리스트 안에 딕셔너리**가 든 모양이 많습니다(학생 명단 등). 각 딕셔너리를 꺼내 `["키"]` 로 값을 읽습니다.

````python
students = [
    {"name": "김민준", "score": 88},
    {"name": "이서윤", "score": 95},
    {"name": "박지호", "score": 72},
    {"name": "최아린", "score": 61},
]
passed = []
failed = []
for student in students:
    if student["score"] >= 80:
        passed.append(student["name"])
    else:
        failed.append(student["name"])
print("합격:", passed)
print("불합격:", failed)
````

**등급별 그룹 묶기** — 딕셔너리의 값으로 리스트를 두면 학생을 그룹으로 묶을 수 있습니다.

````python
students = [
    {"name": "김민준", "score": 88}, {"name": "이서윤", "score": 95},
    {"name": "박지호", "score": 72}, {"name": "최아린", "score": 61},
]
groups = {"상": [], "중": [], "하": []}
for student in students:
    score = student["score"]
    name = student["name"]
    if score >= 90:
        groups["상"].append(name)
    elif score >= 75:
        groups["중"].append(name)
    else:
        groups["하"].append(name)
for level, names in groups.items():
    print(f"{level}: {names}")
````

## 9. 종합 예제 — 상품 재고 관리

**재고 딕셔너리**와 **주문 목록**을 함께 처리합니다. 배운 것(딕셔너리·`for`·`if`·`in`)을 모두 엮습니다.

````python
inventory = {"볼펜": 50, "공책": 30, "색연필": 20, "자": 15}
purchases = [["볼펜", 5], ["공책", 10], ["색연필", 25], ["가위", 3]]

for item, qty in purchases:
    if item not in inventory:
        print(f"{item}: 취급하지 않는 상품")
    elif inventory[item] < qty:
        print(f"{item}: 재고 부족 (현재 {inventory[item]}개)")
    else:
        inventory[item] -= qty
        print(f"{item} {qty}개 판매 완료")

print("--- 남은 재고 ---")
for item, qty in inventory.items():
    print(f"{item}: {qty}개")
````
