## 1. 실수 다루기: `float()`

`int()` 가 문자열을 정수로 바꾸듯, **`float()`** 는 문자열을 **실수(소수)**로 바꿉니다.

````python
print(int("3"))       # 3   (정수)
print(float("3"))     # 3.0 (실수)
print(float("3.14"))  # 3.14
````

소수가 포함된 값을 입력받을 때는 `float()` 로 바꿉니다. 입력 문장과 변환 문장을 두 줄로 나눠 씁니다.

````python
r = input("반지름을 입력하시오: ")
r = float(r)
print(2 * 3.141592 * r)   # 원의 둘레
````

## 2. f-문자열

문자열 앞에 **`f`** 를 붙이고 **`{}`** 안에 변수·수식을 넣으면 그 값이 자동으로 채워집니다.

````python
name = "홍길동"
score = 90
print(f"{name}님의 점수는 {score}점입니다.")
````

`{}` 안에는 수식도 쓸 수 있습니다.

````python
x = 100
y = 200
print(f"{x}과 {y}의 합은 {x + y}, 곱은 {x * y}입니다.")
````

`{값:.nf}` 로 **소수점 아래 자릿수**를 지정합니다. `.2f` 는 둘째 자리까지.

````python
pi = 3.141592653589793
print(f"파이 ≈ {pi:.2f}")
print(f"파이 ≈ {pi:.4f}")
````

**반올림 `round()`** — `{:.2f}` 는 보이는 모양만 바꾸지만, `round()` 는 값 자체를 반올림합니다.

````python
pi = 3.141592653589793
print(round(pi))       # 3
print(round(pi, 2))    # 3.14
print(round(pi, 4))    # 3.1416
````

## 3. 산술 연산자

나눗셈은 두 종류입니다: `/`(실수 나눗셈), `//`(정수 몫). 그리고 `%`(나머지), `**`(지수).

````python
print(7 / 4)    # 1.75 (실수)
print(7 // 4)   # 1    (몫)
print(7 % 4)    # 3    (나머지)
print(2 ** 7)   # 128  (2의 7제곱)
````

`//` 와 `%` 를 함께 쓰면 초를 분과 초로 바꿀 수 있습니다.

````python
sec = 1000
minutes = sec // 60
remainder = sec % 60
print(f"{sec}초는 {minutes}분 {remainder}초입니다.")
````

**문자열에도 쓰이는 `+` 와 `*`** — 같은 연산자라도 자료형에 따라 다르게 동작합니다. `+`는 이어 붙이기, `*`는 반복.

````python
print("안녕" + "하세요")   # 연결
print("=" * 20)            # 20번 반복
print("파이썬! " * 3)       # 3번 반복
````

**연산 결과의 자료형** — `/`는 항상 실수(float), `//`는 정수(int). `type()`으로 확인해 봅니다.

````python
print(type(7 / 4))    # <class 'float'>
print(type(7 // 4))   # <class 'int'>
````

## 4. 복합 대입 연산자와 우선순위

**복합 대입 연산자**는 `num = num + 2` 를 `num += 2` 로 줄여 씁니다.

````python
x = 1000
x += 200   # x = x + 200
print(x)
x -= 100   # x = x - 100
print(x)
x *= 2     # x = x * 2
print(x)
````

> ⚠️ **우선순위**: `*`, `/` 가 `+`, `-` 보다 먼저 계산됩니다. 순서를 바꾸려면 괄호 `()` 를 씁니다.

````python
x, y, z = 80, 90, 70
print(f"잘못된 평균: {x + y + z / 3}")
print(f"올바른 평균: {(x + y + z) / 3}")
````

## 5. 종합 활용

지금까지 배운 것을 엮어 실생활 계산 프로그램을 만듭니다.

**화씨 → 섭씨 변환** (섭씨 = (화씨 − 32) × 5 / 9)

````python
f = input("화씨온도를 입력하시오: ")
f = float(f)
c = (f - 32) * 5 / 9
print(f"화씨 {f}도는 섭씨 {c:.1f}도입니다.")
````

**두 수의 사칙연산 계산기** — 두 정수를 입력받아 여러 연산 결과를 한 번에 출력합니다. `7` 과 `4` 를 입력해 보세요.

````python
a = input("첫째 수: ")
a = int(a)
b = input("둘째 수: ")
b = int(b)
print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")
print(f"{a} / {b} = {a / b:.2f}")
print(f"{a} // {b} = {a // b}  ... 나머지 {a % b}")
````

**자판기 거스름돈** — 몫(`//`)으로 동전 개수, 나머지(`%`)로 남은 돈. `1280` 을 입력해 보세요.

````python
money = input("거스름돈 금액을 입력하시오: ")
money = int(money)

c500 = money // 500
money = money % 500
c100 = money // 100
money = money % 100
c50 = money // 50
money = money % 50
c10 = money // 10

print(f"500원 {c500}개, 100원 {c100}개, 50원 {c50}개, 10원 {c10}개")
````
