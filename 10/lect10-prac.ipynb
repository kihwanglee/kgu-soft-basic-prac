## 1. 리스트 안의 딕셔너리

현실의 데이터는 **리스트 안에 딕셔너리**가 든 모양이 많습니다(학생 명단 등). 각 요소가 딕셔너리이므로, 바깥(리스트)에서 하나를 꺼낸 뒤 안쪽(딕셔너리)에서 `["키"]`로 값을 읽습니다.

````python
students = [
    {"name": "김민준", "score": 88},
    {"name": "이서윤", "score": 95},
    {"name": "박지호", "score": 72},
    {"name": "최아린", "score": 61},
]

# 첫 번째 학생의 이름과 점수
print(students[0]["name"])
print(students[0]["score"])
````

> 💡 `students` → 리스트, `students[0]` → 첫 번째 딕셔너리, `students[0]["name"]` → 그 안의 값. 바깥부터 안쪽으로 한 겹씩 들어갑니다.

````python
# for 문으로 전체를 순회하고, if 문으로 걸러낸다
passed = []
failed = []
for student in students:
    if student["score"] >= 80:
        passed.append(student["name"])
    else:
        failed.append(student["name"])

print("합격:", passed)
print("불합격:", failed)
````

## 2. 등급별로 그룹 묶기

딕셔너리의 **값을 리스트로** 두면 항목을 그룹별로 모을 수 있습니다. 등급(A~D)을 키로 두고, 각 값을 빈 리스트로 시작해 이름을 `append` 합니다.

````python
students = [
    {"name": "김민준", "score": 88}, {"name": "이서윤", "score": 95},
    {"name": "박지호", "score": 72}, {"name": "최아린", "score": 61},
    {"name": "정하준", "score": 83}, {"name": "윤지수", "score": 91},
]

groups = {"A": [], "B": [], "C": [], "D": []}
for student in students:
    score = student["score"]
    name = student["name"]
    if score >= 90:
        groups["A"].append(name)
    elif score >= 80:
        groups["B"].append(name)
    elif score >= 70:
        groups["C"].append(name)
    else:
        groups["D"].append(name)

for grade, names in groups.items():
    if len(names) > 0:        # 아무도 없는 등급은 건너뛴다
        print(f"{grade}등급: {names}")
````

## 3. 종합 예제: 상품 재고 관리

**재고 딕셔너리**와 **주문 목록(리스트 안의 리스트)**을 함께 처리합니다. 딕셔너리·`for`·`if`·`in`을 모두 엮는 문제입니다.

````python
inventory = {"볼펜": 50, "공책": 30, "색연필": 20, "자": 15}
purchases = [["볼펜", 5], ["공책", 10], ["색연필", 25], ["가위", 3]]

print("=== 구매 처리 ===")
for item, qty in purchases:
    if item not in inventory:
        print(f"{item}: 취급하지 않는 상품")
    elif inventory[item] < qty:
        print(f"{item}: 재고 부족 (현재 {inventory[item]}개)")
    else:
        inventory[item] -= qty
        print(f"{item} {qty}개 판매 완료")

print("--- 남은 재고 ---")
for item, qty in inventory.items():
    print(f"{item}: {qty}개")
````

## 4. 리스트를 포함한 리스트 (2차원 표)

리스트 안에 리스트가 든 구조로 **행과 열이 있는 표**를 저장합니다. `표[행][열]` 로 접근하며 인덱스는 0부터 시작합니다.

````python
# 3명의 학생, 3과목 점수 (행: 학생, 열: 국어·수학·영어)
scores = [
    [85, 92, 78],   # 1번 학생
    [90, 88, 95],   # 2번 학생
    [72, 80, 68],   # 3번 학생
]

# 2번 학생(인덱스 1)의 수학 점수(인덱스 1)
print(scores[1][1])
````

````python
# 학생별 총점·평균
for i in range(len(scores)):
    row = scores[i]
    total = row[0] + row[1] + row[2]
    print(f"{i + 1}번 학생 총점 {total}, 평균 {total // 3}")
````

> ⚠️ **주의**: `scores[0]` 은 첫 번째 행 전체, `scores[0][2]` 는 첫 번째 행의 세 번째 값입니다.

## 6. `while` 반복문

`for` 문이 **정해진 횟수**만큼 반복한다면, `while` 문은 **조건이 참인 동안** 계속 반복합니다.

```
while 조건식:
    반복할 코드
```

````python
i = 1
while i <= 5:
    print(i)
    i += 1    # 이 줄이 없으면 무한 루프!
````

> ⚠️ **주의: 무한 루프** — 조건이 항상 참이면 루프가 끝나지 않습니다. 반복 **안에서** 조건에 쓰인 변수(`i`)를 반드시 바꿔, 언젠가 조건이 거짓이 되게 해야 합니다.
>
> 실행이 멈추지 않으면 코랩의 **중지(■)** 버튼을 누르세요.

````python
# for 문으로 쓴 같은 결과
for i in range(1, 6):
    print(i)
````

## 7. `while` 활용: 합산과 카운트다운

````python
# 카운트다운
count = 5
while count > 0:
    print(count)
    count -= 1
print("발사!")
````

## 8. `break` 와 `continue`

- `break`: 루프를 **완전히 종료**하고 빠져나갑니다.
- `continue`: 현재 반복만 건너뛰고 **다음 반복**으로 넘어갑니다.

````python
# break: 포도를 찾으면 멈춘다
fruits = ["사과", "바나나", "포도", "딸기", "수박"]
for fruit in fruits:
    if fruit == "포도":
        print("포도를 찾았습니다!")
        break
    print(f"{fruit} 검사 중...")
````

````python
# continue: 홀수는 건너뛰고 짝수만 출력
for i in range(1, 11):
    if i % 2 != 0:
        continue
    print(i)
````

## 9. `while True` + `break`: 입력 반복과 게임

종료 조건이 복잡하거나 **입력을 반복해서 받아야 할 때** `while True:` 로 무한 반복하다가 `break` 로 빠져나옵니다.

> 아래 두 셀은 실행하면 **입력창**이 뜹니다. 안내에 따라 값을 입력해 보세요.

````python
# 점수를 여러 번 입력받아 합산 (-1을 입력하면 종료)
total = 0
while True:
    score = input("점수를 입력하시오 (종료: -1): ")
    score = int(score)
    if score == -1:
        break
    total += score
print(f"합계: {total}")
````

````python
# 숫자 맞히기 게임
import random

answer = random.randint(1, 20)   # 1~20 중 무작위 정수
tries = 0
print("1~20 사이의 수를 맞혀보세요.")

while True:
    guess = input("추측: ")
    guess = int(guess)
    tries += 1
    if guess == answer:
        print(f"정답! {tries}번 만에 맞혔습니다.")
        break
    elif guess < answer:
        print("더 큰 수입니다.")
    else:
        print("더 작은 수입니다.")
````

> 💡 `random`은 파이썬 기본 모듈이라 설치 없이 `import random` 한 줄만 적으면 됩니다.

## 10. 종합 예제: 주문 입력받아 집계하기

이번 시간의 두 축인 **`while`(반복 입력)**과 **딕셔너리(빈도 세기)**를 함께 씁니다. `끝`을 입력할 때까지 메뉴를 받아 메뉴별 주문 수를 셉니다.

````python
count = {}
while True:
    menu = input("주문 메뉴 (종료: 끝): ")
    if menu == "끝":
        break
    if menu in count:
        count[menu] += 1
    else:
        count[menu] = 1

print("=== 메뉴별 주문 수 ===")
for menu, num in count.items():
    print(f"{menu}: {num}잔")
````
