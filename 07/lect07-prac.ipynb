## 1. 리스트와 `for` 종합 (누적)

반복 **밖**에서 그릇(`total = 0`)을 준비하고, 반복 **안**에서 더하거나 세면 합계·개수를 구할 수 있습니다.

**성적표 분석** — 합계·평균·최고/최저·합격자 수를 한 번의 반복에서 구합니다.

````python
scores = [88, 95, 70, 100, 64, 82]
total = 0
pass_count = 0
for score in scores:
    total += score
    if score >= 70:
        pass_count += 1
print(f"학생 수: {len(scores)}명")
print(f"평균: {total / len(scores):.1f}점")
print(f"최고점: {max(scores)}, 최저점: {min(scores)}")
print(f"70점 이상: {pass_count}명")
````

**가장 큰 값 직접 찾기** — `max()` 없이, 첫 값을 후보로 두고 더 큰 값을 만나면 바꿉니다.

````python
scores = [88, 95, 70, 100, 64, 82]
biggest = scores[0]
for score in scores:
    if score > biggest:
        biggest = score
print(biggest)   # 100
````

## 2. 중첩 `for` 문과 구구단

`for` 안에 `for` 를 넣으면, 바깥이 한 번 돌 때 안쪽이 **처음부터 끝까지** 돕니다.

````python
for i in range(1, 4):       # 바깥: 1, 2, 3
    for j in range(1, 4):   # 안쪽: 1, 2, 3
        print(i, j)
````

**사각형** — 안쪽으로 한 줄의 별, 바깥으로 줄 수. `end=""`는 줄을 바꾸지 않습니다.

````python
for row in range(3):
    for col in range(5):
        print("*", end="")
    print()
````

**구구단 2단** — 먼저 한 단만.

````python
dan = 2
for i in range(1, 10):
    print(f"{dan} x {i} = {dan * i}")
````

## 3. 딕셔너리 — 이름표로 값 찾기

리스트는 **번호**로, 딕셔너리는 **이름(키)**으로 값을 찾습니다. `{키: 값, ...}` 형태입니다.

````python
prices = {"사과": 1000, "바나나": 1500, "포도": 3000}
print(prices["사과"])     # 1000  (키로 값 꺼내기)
````

**추가·수정** — `딕셔너리[키] = 값`. **확인·개수** — `in`, `len()`.

````python
prices = {"사과": 1000, "바나나": 1500}
prices["포도"] = 3000     # 새 키 추가
prices["사과"] = 1200     # 기존 값 수정
print(prices)
print("사과" in prices, len(prices))   # True 3
````

**`for` 로 다루기** — 키를 하나씩, 또는 `.items()`로 키·값을 함께.

````python
prices = {"사과": 1000, "바나나": 1500, "포도": 3000}
for fruit in prices:
    print(f"{fruit}: {prices[fruit]}원")

for fruit, price in prices.items():
    print(f"{fruit}는 {price}원입니다.")
````

**글자 개수 세기** — 딕셔너리는 "무엇이 몇 개"를 셀 때 편합니다.

````python
word = "banana"
count = {}
for ch in word:
    if ch in count:
        count[ch] += 1
    else:
        count[ch] = 1
print(count)   # {'b': 1, 'a': 3, 'n': 2}
````
