## 1. 무한 루프 피하기

조건이 **영원히 참**이면 루프가 끝나지 않습니다. 흔한 원인은 세 가지입니다. (아래 코드는 실행하지 말고 눈으로만 보세요.)

```python
# ① 변경을 빠뜨렸다
i = 1
while i <= 5:
    print(i)          # i += 1 이 없다

# ② 반대 방향으로 변경했다
i = 1
while i <= 5:
    print(i)
    i -= 1            # i가 점점 작아진다

# ③ 조건이 정확히 맞아떨어지지 않는다
i = 1
while i != 10:        # 1, 3, 5, 7, 9, 11, ... 10이 되는 순간이 없다
    print(i)
    i += 2
```

> ⚠️ **무한 루프에 빠졌다면** 셀 왼쪽의 **중지(■)** 버튼을 누르세요. 화면이 멈추면 **런타임 → 세션 다시 시작**을 누릅니다.

③은 `!=` 대신 **범위로 검사**(`<`)하면 해결됩니다.

````python
i = 1
while i < 10:         # "같지 않은 동안" 대신 "작은 동안"
    print(i)
    i += 2
````

## 3. 입력을 반복해 받기: 세 단계

`while` 문이 가장 빛나는 곳은 **사용자 입력을 반복해서 받는** 경우입니다. 몇 번 입력할지 미리 알 수 없기 때문입니다. `while True:`와 `break`를 **세 단계로 차근차근** 익혀 봅시다.

> 이 절의 코드 셀은 실행하면 **입력창**이 뜹니다. 안내에 따라 값을 입력해 보세요.

**1단계: `while True:`와 `break`로 조건만 검사하기.** 무한히 반복하되, 입력을 받아 `if`로 조건을 검사해 맞으면 `break`로 빠져나옵니다. 더하거나 세는 것 없이 **검사와 종료만** 합니다.

````python
while True:
    password = input("비밀번호를 입력하시오: ")
    if password == "python":
        print("로그인 성공!")
        break
    print("비밀번호가 틀렸습니다. 다시 입력하세요.")
````

**2단계: 초기값은 밖에서, 변경은 입력으로 (입력이 두 번).** 들어온 값을 차례로 더하고 `0`이 들어오면 멈춥니다. `while 조건식:`으로 쓰면 조건 변수(`amount`)를 **반복 전에 한 번**, **반복 안에서 다시** 입력받습니다. 그래서 입력 코드가 두 번 나옵니다.

````python
total = 0
amount = input("지출 금액 (종료: 0): ")     # ① 초기값: 반복 밖에서 한 번 입력
amount = int(amount)

while amount != 0:                           # ② 조건 검사
    total += amount
    amount = input("지출 금액 (종료: 0): ")   # ③ 변경: 반복 안에서 다시 입력
    amount = int(amount)

print(f"총 지출: {total}원")
````

**3단계: `while True:` + `break`로 입력을 한 번만.** 2단계와 똑같은 일을, 입력 코드를 **한 번만** 써서 합니다. 입력도 종료 검사도 모두 반복 안에서 하므로 중복이 사라집니다. 입력 반복에는 보통 이 방식이 편합니다.

````python
total = 0
while True:
    amount = input("지출 금액 (종료: 0): ")
    amount = int(amount)
    if amount == 0:
        break
    total += amount
print(f"총 지출: {total}원")
````

**올바른 값이 들어올 때까지 다시 묻기(입력 검증):** 잘못된 값인 동안 계속 다시 묻습니다. 2단계처럼 반복 전에 한 번, 반복 안에서 한 번 입력받는 형태입니다.

````python
score = input("만족도를 입력하시오 (1~5): ")
score = int(score)

while score < 1 or score > 5:
    print("1부터 5 사이의 수를 입력하세요.")
    score = input("만족도를 입력하시오 (1~5): ")
    score = int(score)

print(f"만족도 {score}점이 기록되었습니다.")
````

**메뉴를 반복해서 보여 주기:** 메뉴 번호는 `int()`로 바꾸지 않고 **문자열로 비교**하면 글자를 잘못 입력해도 오류가 나지 않습니다.

````python
balance = 0

while True:
    print("1. 입금  2. 출금  3. 잔액 조회  4. 종료")
    choice = input("번호를 선택하시오: ")

    if choice == "1":
        amount = input("입금액: ")
        amount = int(amount)
        balance += amount
    elif choice == "2":
        amount = input("출금액: ")
        amount = int(amount)
        if amount > balance:
            print("잔액이 부족합니다.")
        else:
            balance -= amount
    elif choice == "3":
        print(f"현재 잔액: {balance}원")
    elif choice == "4":
        print("이용해 주셔서 감사합니다.")
        break
    else:
        print("잘못된 번호입니다.")
````

## 4. `break`와 `continue`

`break`는 `while True:`에서만 쓰는 게 아닙니다. **`for` 문에서도** 조건을 만나면 바로 멈출 수 있습니다. 월별 이익 목록에서 **처음으로 적자(음수)**가 난 달을 찾습니다. 찾으면 더 볼 필요가 없으니 `break`.

````python
profits = [120, 80, -30, 50, -10]
for i in range(len(profits)):
    if profits[i] < 0:
        print(f"{i + 1}월에 처음 적자가 났다: {profits[i]}")
        break
````

5월의 `-10`은 검사하지 않습니다. `break`로 이미 루프를 빠져나왔기 때문입니다. 같은 일을 `while` 문으로도 쓸 수 있지만, 그때는 리스트 범위를 벗어나지 않게 `i < len(profits)` 조건을 직접 챙겨야 합니다. `for` 문은 그 걱정이 없어 더 간단합니다.

> 💡 `break`는 `for`·`while` 어디서나 **가장 가까운 반복을 즉시 끝냅니다.** "찾으면 멈추기"에 씁니다.

`continue`는 **이번 반복만 건너뛰고** 다음 반복으로 넘어갑니다. `for` 문에서 가장 자연스럽습니다. 아래는 1~10 중 홀수만 출력합니다.

````python
for n in range(1, 11):
    if n % 2 == 0:
        continue       # 짝수는 건너뛴다
    print(n)           # 홀수만 출력된다
````

> ⚠️ `while` 문에서 `continue`를 쓰면 **변경 코드까지 건너뛰어 무한 루프**가 되기 쉽습니다. 변경 코드를 `continue`보다 **앞**에 두어야 안전합니다. (이번 학기에는 `continue`를 `for` 문에서만 써도 충분합니다.)

## 5. `while`과 리스트: 인덱스로 순회하기

리스트의 항목을 처음부터 끝까지 하나씩 보려면 **위치를 가리키는 번호(인덱스)**를 `0`에서 시작해 하나씩 늘립니다. `while` 문의 조건은 `인덱스 < len(리스트)`로 둡니다. 인덱스는 0부터지만 **`i + 1`로 "1번, 2번, 3번"**처럼 순서를 붙일 수 있습니다.

````python
fruits = ["사과", "바나나", "포도", "딸기"]
i = 0
while i < len(fruits):
    print(f"{i + 1}번: {fruits[i]}")
    i += 1
````

같은 일을 `for` 문으로 쓰면 `range(len(...))`가 인덱스를 대신 만들어 줍니다. 초기화·조건·변경을 직접 쓰지 않아 더 간결합니다. 리스트를 처음부터 끝까지 훑는 일은 이렇게 `for` 문이 더 간단합니다.

````python
fruits = ["사과", "바나나", "포도", "딸기"]
for i in range(len(fruits)):
    print(f"{i + 1}번: {fruits[i]}")
````

## 6. 종합 예제: 기회가 정해진 숫자 맞히기

> 아래 셀은 실행하면 **입력창**이 뜹니다.

`success`처럼 참·거짓 상태를 기억하는 변수를 두면, 루프가 끝난 뒤 **왜 끝났는지**(맞혀서인지 기회를 다 써서인지) 구분할 수 있습니다.

````python
import random

answer = random.randint(1, 50)
chance = 5
success = False      # 맞혔는지 기억하는 변수

while chance > 0 and not success:
    guess = input(f"1~50 사이의 수 (남은 기회 {chance}번): ")
    guess = int(guess)
    chance -= 1

    if guess == answer:
        success = True
    elif guess < answer:
        print("더 큰 수입니다.")
    else:
        print("더 작은 수입니다.")

if success:
    print("정답입니다!")
else:
    print(f"기회를 모두 썼습니다. 정답은 {answer}였습니다.")
````

## 7. 함수 정의·호출과 매개변수

`def 이름():` 로 함수를 **정의**하고 `이름()` 으로 **호출**합니다. 정의만으로는 실행되지 않고, 호출해야 몸체가 실행됩니다. 괄호 안에 **매개변수**를 두면 값을 건네받아 다르게 동작합니다.

````python
def greet():
    print("안녕하세요!")
    print("반갑습니다.")

greet()      # 호출해야 실행된다
greet()      # 여러 번 부를 수 있다
````

````python
def greet(name):
    print(f"{name}님, 안녕하세요!")

greet("지민")      # 인자 '지민'이 매개변수 name으로 전달
greet("서준")
````

````python
def introduce(name, age):
    print(f"저는 {name}이고, {age}살입니다.")

introduce("하늘", 20)      # 인자를 순서대로 전달
````

## 8. `return` 으로 값 돌려주기

`return` 은 계산 결과를 **호출한 곳으로 돌려줍니다.** `print` 는 화면에 보여 줄 뿐이지만, `return` 으로 돌려받은 값은 변수에 담거나 다른 계산에 **다시 쓸 수 있습니다.**

````python
def add(a, b):
    result = a + b
    return result

total = add(3, 5)
print(total)       # 8
````

````python
def square(n):
    return n * n

print(square(4))                # 16
print(square(4) + square(3))    # 25 (돌려받은 값을 바로 계산에 사용)
````

## 9. 조건·반복과 함께 쓰는 함수

함수 몸체 안에서 `if`·`for`·`while` 을 자유롭게 씁니다. 자주 하는 판단이나 반복을 함수로 묶으면 **이름만 바꿔** 여러 번 재사용할 수 있습니다.

````python
def grade(score):
    if score >= 60:
        return "합격"
    else:
        return "불합격"

print(grade(75))      # 합격
print(grade(40))      # 불합격
````

````python
def range_sum(end):
    total = 0
    for i in range(1, end + 1):
        total += i
    return total

print(range_sum(10))      # 55
print(range_sum(100))     # 5050
````

리스트를 받아 그 안 숫자들의 **평균**을 돌려주는 함수 `average(numbers)` 를 만드세요. `average([80, 90, 100])` 은 `90.0` 을 돌려줘야 합니다. (힌트: `sum(numbers)`, `len(numbers)`)

<details><summary>💡 해답 보기</summary>


`sum` 으로 합, `len` 으로 개수를 구해 나눕니다. 둘 다 파이썬이 만들어 둔 함수입니다.

## 정리

- **함수 정의·호출**: `def 이름():` 로 정의하고 `이름()` 으로 호출합니다. 정의만으로는 실행되지 않습니다.
- **매개변수와 인자**: 괄호 안 매개변수로 값을 건네고, 호출할 때 인자를 순서대로 넣습니다.
- **`return`**: 결과를 호출한 곳으로 돌려줍니다. `print` 는 보여 줄 뿐, `return` 은 다시 쓸 값을 돌려줍니다.
- **왜 함수를 쓰나**: 반복을 줄이고, 고칠 곳을 한 군데로 모으고, 이름으로 뜻을 분명히 합니다. 조건문·반복문을 함수로 묶어 재사용합니다.
