## 개념: 변수는 '이름'과 '값'의 연결

컴퓨터 세계에서 어떤 대상을 다루려면 **이름(name)·값(value)·유형(type)**이 정해져야 합니다. 그런데 **유형은 값에 딸린 속성**(예: `90`은 정수, `"홍길동"`은 문자열)이라, 우리가 실제로 하는 일은 **이름에 값을 연결(할당)**하는 것입니다.

자연언어의 **단어**가 **형식(소리·글자)**과 **의미**가 짝을 이뤄야 유효하듯, 변수도 **이름(형식)**에 **값(의미)**이 연결돼야 유효합니다.

> **형식(소리·글자) ↔ 의미**  ≈  **이름(name) ↔ 값(value)**

| 자연언어의 단어 | 파이썬의 변수 |
|:---:|:---:|
| 형식(소리·글자) | 이름(name) |
| 의미 | 값(value) |
| 형식과 의미가 짝이어야 단어 | 이름에 값이 연결돼야 변수 |

값을 연결한 적 없는 이름을 쓰면 `NameError`(예: `name 'nickname' is not defined`)가 납니다. "의미(값)가 없는 이름"이라 파이썬이 알아듣지 못한다는 뜻입니다.

## 1. 변수의 이름과 종류

변수 이름(식별자)은 규칙을 지켜 짓습니다.
- **의미 있는 이름**을 쓰고, **숫자로 시작할 수 없으며**, 이름 중간에 **공백이 들어갈 수 없습니다.**
- **대문자와 소문자는 다른 이름**입니다(`name` ≠ `Name`).
- 여러 단어는 밑줄로 잇는 **뱀체(snake_case)**를 씁니다. 예: `my_score`

> ⚠️ **주의**: `2nd_base`(숫자로 시작), `money#`(기호 사용)는 변수 이름으로 쓸 수 없습니다.

아래는 규칙에 맞는 이름으로 값을 저장하는 예입니다. 실행해 보세요.

````python
my_score = 90
number_of_pictures = 12
print(my_score)
print(number_of_pictures)
````

변수에는 새 값을 다시 저장할 수 있고, **자기 자신을 이용해** 값을 바꿀 수도 있습니다.

````python
score = 10
score = score + 1
print(score)   # 11
````

## 2. 문자열 변수 활용

`print()` 에 쉼표(`,`)로 값을 구분하면 변수와 글자를 섞어 한 문장처럼 출력할 수 있습니다.

````python
x = 100
y = 200
total = x + y
print(x, "과", y, "의 합은", total, "입니다.")
````

문자열 변수는 `+` 로 이어 붙여(연결) 원하는 문장을 만들 수 있습니다.

````python
name = "홍길동"
dept = "소프트웨어경영학과"
print("이름: " + name)
print("학과: " + dept)
print("환영합니다, " + name + "님!")
````

## 3. 사용자로부터 입력 받기: `input()`

`input("안내 문구")` 는 안내 문구를 보여 주고, 사용자가 입력한 값을 **문자열**로 돌려줍니다.

> 실행하면 입력창이 나타납니다. 값을 입력하고 `Enter` 를 누르세요.

````python
name = input("이름을 입력하시오: ")
print(name, "씨, 안녕하세요?")
print("파이썬에 오신 것을 환영합니다.")
````

### 정수 입력 받기

`input()` 이 돌려주는 값은 **항상 문자열**입니다. 계산에 쓰려면 `int()` 로 정수로 바꿔야 합니다. 읽기 쉽도록 **입력 문장**과 **변환 문장**을 두 줄로 나눠 씁니다.

````python
x = input("첫 번째 정수를 입력하시오: ")
x = int(x)
y = input("두 번째 정수를 입력하시오: ")
y = int(y)
total = x + y
print(x, "과", y, "의 합은", total, "입니다.")
````

> ⚠️ **주의**: `int()` 로 바꾸지 않으면 `"3" + "5"` 처럼 **문자열 연결**이 되어 `35` 가 나옵니다. 숫자로 바꿔야 `3 + 5 = 8` 로 계산됩니다.

**나이 계산기** — 태어난 해를 입력받아 올해(2026년) 기준 나이를 구합니다.

````python
year = input("태어난 해를 입력하시오: ")
year = int(year)
age = 2026 - year
print("당신의 나이는", age, "살입니다.")
````

## 4. 종합 예제: 간단한 계산기

**입력·변수·계산·출력**을 모두 사용해, 두 수를 입력받아 사칙연산 결과를 한꺼번에 보여 줍니다. `10` 과 `3` 을 입력해 보세요.

````python
a = input("첫 번째 수를 입력하시오: ")
a = int(a)
b = input("두 번째 수를 입력하시오: ")
b = int(b)

print(a, "+", b, "=", a + b)
print(a, "-", b, "=", a - b)
print(a, "*", b, "=", a * b)
print(a, "/", b, "=", a / b)
````

**원 그리기** — 반지름을 변수 `radius` 에 저장했습니다. 실행하면 원 3개가 그려집니다.

````python
from ColabTurtlePlus.Turtle import clearscreen, circle, forward
clearscreen()

radius = 100
circle(radius)
forward(30)
circle(radius)
forward(30)
circle(radius)
````

**변수의 유용성** — 위 코드에서 `radius` 값만 `50` 으로 바꿔 다시 실행해 보세요. 세 원의 크기가 **모두 함께** 바뀝니다. 값을 한 곳에서 관리할 수 있는 것이 변수의 큰 장점입니다.

**집 그리기** — 사용자에게 집의 크기를 입력받아 집을 그립니다.

````python
from ColabTurtlePlus.Turtle import clearscreen, forward, left, right

size = input("집의 크기는 얼마로 할까요? ")
size = int(size)

clearscreen()
left(60)
forward(size)    # 1
right(120)
forward(size)    # 2
right(30)
forward(size)    # 3
right(90)
forward(size)    # 4
right(90)
forward(size)    # 5
right(90)
forward(size)    # 6
````
