## 1. 코랩 노트북은 '셀'로 이루어져 있어요

코랩 노트북은 위에서 아래로 이어지는 **셀(cell)** 들의 묶음입니다. 셀에는 두 종류가 있습니다.

| 셀 종류 | 하는 일 | 지금 이 셀은? |
|:---:|:---|:---|
| **텍스트 셀** | 제목·설명을 글로 씁니다 (마크다운) | 텍스트 셀입니다 |
| **코드 셀** | 파이썬 코드를 입력하고 **실행**합니다 | 아래에 있어요 ↓ |

- 셀을 한 번 클릭하면 **선택**, 두 번 클릭하면 **편집**할 수 있습니다.
- 새 셀은 위쪽 메뉴의 **[+ 코드]**, **[+ 텍스트]** 버튼으로 추가합니다.

### 셀을 실행하는 방법

아래 코드 셀을 클릭한 뒤, 다음 중 하나로 실행해 보세요.

- 셀 **왼쪽의 실행 버튼(▶)** 을 클릭
- 키보드로 **`Shift` + `Enter`** (실행하고 다음 셀로 이동)

> 💡 처음 실행하면 코랩이 클라우드에 파이썬 환경(런타임)을 준비하느라 **몇 초** 걸릴 수 있어요. 기다리면 됩니다.

````python
# 이 셀을 클릭하고 Shift+Enter 를 눌러 실행해 보세요.
print("Hello, World!")
````

방금 무슨 일이 일어났나요?

- `print(...)` 은 괄호 안의 내용을 **화면에 출력**하는 명령입니다.
- 셀 **아래쪽에 결과** `Hello, World!` 가 나타났다면 성공입니다. 🎉

## 2. `print()` 으로 화면에 출력하기

`print()` 는 이 수업에서 가장 많이 쓰게 될 명령입니다. 괄호 안에 출력할 내용을 넣습니다.

- 글자(문자열)를 출력할 때는 반드시 **따옴표**로 감쌉니다. `"..."` 또는 `'...'` 둘 다 됩니다.
- 아래 셀을 실행해 보세요.

````python
print("안녕하세요!")
print('반갑습니다.')
````

- 큰따옴표 `"..."` 와 작은따옴표 `'...'` 는 결과가 같습니다. 편한 쪽을 쓰면 됩니다.
- `print()` 를 여러 번 쓰면 **한 줄씩** 차례로 출력됩니다.

### 값을 여러 개 함께 출력하기

`print()` 안에서 값을 **쉼표(`,`)** 로 구분하면, 값들 사이에 공백을 두고 함께 출력합니다.

````python
print("결과값은", 2 * 7, "입니다.")
````

- 따옴표로 감싼 `"결과값은"` 은 글자 그대로 출력되고,
- 따옴표가 없는 `2 * 7` 은 **계산되어** `14` 로 출력됩니다.
- 즉 파이썬은 **글자**와 **계산할 값**을 구분합니다. 이 차이를 기억해 두세요.

## 3. 파이썬을 계산기처럼 써 보기

코드 셀에 수식을 넣고 실행하면 파이썬이 **계산**해 줍니다. 먼저 아래를 실행해 보세요.

````python
print(3 + 2)
print(2 ** 10)
````

자주 쓰는 산술 연산자입니다. (자세한 내용은 다음 시간에 배웁니다.)

| 연산자 | 의미 | 예시 | 결과 |
|:---:|:---:|:---:|:---:|
| `+` `-` `*` | 덧셈·뺄셈·곱셈 | `3 * 4` | `12` |
| `/` | 나눗셈(실수) | `7 / 2` | `3.5` |
| `**` | 거듭제곱 | `2 ** 10` | `1024` |

아래 셀에서 나눗셈과 거듭제곱도 확인해 보세요.

````python
print(7 / 2)
print(5 ** 3)
````

## 4. 파이썬 맛보기: 문장처럼 읽히는 코드

파이썬은 문법이 영어 문장에 가까워 읽기 쉽습니다. 아래 코드는 "사과가 목록 안에 있으면 안내문을 출력하라"는 뜻입니다. 실행해 보세요.

지금은 뜻을 정확히 몰라도 괜찮아요. **눈으로 읽으면 대충 무슨 뜻인지 짐작되는 것** 이 파이썬의 매력입니다.

````python
if "사과" in ["딸기", "바나나", "포도", "사과"]:
    print("사과가 있습니다")
````

- `["딸기", "바나나", "포도", "사과"]` 는 여러 값을 담은 **목록(리스트)** 입니다.
- `in` 은 "~안에 있는가?" 를 확인합니다.
- `if` 는 "조건이 맞으면 아래를 실행" 하라는 뜻입니다.
- 리스트, `in`, `if` 는 앞으로 차근차근 배웁니다. 오늘은 **실행되는 것만 확인** 하면 충분합니다.
