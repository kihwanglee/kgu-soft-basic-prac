## 1. `print()` 으로 출력하기

## 2. 숫자 계산하기

파이썬은 강력한 계산기입니다. 주요 산술 연산자는 다음과 같습니다.

| 연산자 | 의미 | 예 | 결과 |
|:---:|:---:|:---:|:---:|
| `+` `-` `*` | 덧셈·뺄셈·곱셈 | `3 * 4` | `12` |
| `/` | 나눗셈(실수) | `7 / 2` | `3.5` |
| `//` | 몫(정수) | `7 // 2` | `3` |
| `%` | 나머지 | `7 % 2` | `1` |
| `**` | 거듭제곱 | `2 ** 8` | `256` |

````python
print(3 + 2)
print(3 * 4)
print(2 ** 8)
````

나눗셈은 세 종류가 있습니다: `/`(실수 나눗셈), `//`(정수 몫), `%`(나머지).

````python
print(7 / 2)   # 3.5
print(7 // 2)  # 3
print(7 % 2)   # 1
````

> ⚠️ **주의**: `/` 의 결과는 나누어떨어져도 항상 **실수**입니다. `10 / 5` 는 `2` 가 아니라 `2.0`.

````python
print(10 + 5)  # 15 (정수)
print(10 / 5)  # 2.0 (실수)
````

곱셈·나눗셈이 덧셈·뺄셈보다 먼저 계산됩니다. 순서를 바꾸려면 괄호를 씁니다.

````python
print(3 + 2 * 4)    # 11
print((3 + 2) * 4)  # 20
````

## 3. 문자열 다루기

**문자열**은 따옴표로 감싼 텍스트입니다. `+` 로 잇고(연결), `*` 로 반복할 수 있어요.

````python
print("강아지" + "고양이")
print("파이썬" + " " + "재미있다")
````

````python
print("Ha" * 3)
print("-" * 20)
````

같은 `+` 라도 대상이 문자열이면 **연결**, 숫자면 **덧셈**입니다. 문자열과 숫자는 바로 더할 수 없어 `str()` 로 바꿔 줍니다.

````python
print("100" + "200")   # 100200 (연결)
print(100 + 200)       # 300 (덧셈)
print("나이: " + str(20))  # 나이: 20
````

**요약**

| 식 | 대상 | 동작 | 결과 |
|:---:|:---:|:---:|:---:|
| `3 + 2` | 숫자 | 덧셈 | `5` |
| `"3" + "2"` | 문자열 | 연결 | `"32"` |
| `3 * 2` | 숫자 | 곱셈 | `6` |
| `"3" * 2` | 문자열 | 반복 | `"33"` |

## 4. 자료형(data type)

**값의 종류**를 자료형이라고 합니다. 기본 자료형 세 가지:

| 자료형 | 이름 | 예 |
|:---:|:---:|:---|
| 정수 | `int` | `3`, `100` |
| 실수 | `float` | `3.5`, `2.0` |
| 문자열 | `str` | `"안녕"`, `'100'` |

`type()` 함수로 값의 자료형을 확인할 수 있어요.

````python
print(type(100))
print(type(3.5))
print(type("100"))
````

자료형에 따라 `+` 의 동작이 달라지므로, 필요하면 `str()`(→문자열)·`int()`(→정수)로 바꿔 맞춥니다.

````python
print("나이는 " + str(20) + "살")
print(int("15") + int("5"))
````

## 5. 변수

**변수**는 값을 저장해 두는 상자입니다. `=` 는 오른쪽 값을 왼쪽 이름에 **저장**하라는 기호예요 (수학의 '같다'가 아닙니다).

````python
x = 100
print(x)
````

변수에는 새 값을 다시 저장할 수 있고, 그러면 이전 값은 사라집니다.

````python
x = 100
x = 200
print(x)
````

변수 여러 개를 만들어 계산에 사용할 수 있습니다.

````python
x = 100
y = 200
total = x + y
print(total)
````

변수에는 문자열도 저장할 수 있어요.

````python
name = "홍길동"
address = "서울시 종로구 1번지"
print(name)
print(address)
````

> ⚠️ **변수 이름 규칙**: 의미 있는 이름을 쓰고, 숫자로 시작할 수 없으며, 대문자와 소문자는 다른 이름으로 취급됩니다. 여러 단어는 밑줄로 잇습니다(예: `my_score`).

## 6. 종합 예제

지금까지 배운 **변수·계산·문자열**을 함께 써 봅니다. 값을 바꿔 가며 실행해 보세요.

**두 과목 점수의 합과 평균**

````python
korean = 90
english = 80
total = korean + english
average = total / 2
print("총점은", total, "평균은", average)
````

**화씨 온도를 섭씨로 변환** (섭씨 = (화씨 − 32) × 5 / 9)

````python
fahrenheit = 77
celsius = (fahrenheit - 32) * 5 / 9
print("화씨", fahrenheit, "도는 섭씨", celsius, "도이다.")
````

**초를 분과 초로 변환** (`//` 와 `%` 활용)

````python
seconds = 1000
minutes = seconds // 60
rest = seconds % 60
print(seconds, "초는", minutes, "분", rest, "초이다.")
````

## 7. 오류 메시지 읽어 보기

오류는 실패가 아니라 **무엇을 고쳐야 하는지 알려 주는 안내**입니다. 아래 셀들은 **일부러 오류를 내는 코드**를 주석(`#`)으로 막아 두었습니다. 맨 앞의 `#` 을 지우고 실행하여 오류 메시지를 직접 확인해 보세요. 확인한 뒤에는 다시 `#` 을 붙이거나 고쳐 두면 됩니다.

````python
# 따옴표를 빠뜨리면 NameError 가 납니다. 아래 줄의 맨 앞 # 을 지우고 실행해 보세요.
# print(안녕하세요)
print("안녕하세요")   # 이렇게 따옴표로 감싸면 정상 출력됩니다
````

````python
# 문자열과 숫자를 바로 더하면 TypeError 가 납니다. 아래 줄의 # 을 지우고 실행해 보세요.
# print("점수: " + 90)
print("점수: " + str(90))   # str() 로 바꾸면 정상
print("점수:", 90)          # 또는 쉼표로 나누어도 됩니다
````

> 💡 **오류를 만나면**: 메시지의 **마지막 줄**(오류 종류와 설명)과 **오류가 난 줄**을 먼저 확인하세요. 자주 만나는 오류: 따옴표 누락, 문자열과 숫자를 더함, 없는 변수 사용, 괄호 짝이 안 맞음.
