## 1. 관계 연산자

두 값을 비교하여 참(`True`)/거짓(`False`)을 돌려줍니다: `==`, `!=`, `>`, `<`, `>=`, `<=`.

````python
print(7 > 4)    # True
print(7 < 4)    # False
print(5 == 5)   # True
print(5 != 5)   # False
````

문자열도 `==`, `!=`로 비교할 수 있습니다. 대소문자는 구분됩니다.

````python
print("python" == "python")   # True
print("Cat" == "cat")         # False (대소문자 구분)
````

> ⚠️ `=`는 **값을 저장(할당)**, `==`는 **같은지 비교**. 서로 다릅니다!

**산수 퀴즈** — 답이 맞는지 관계 연산자로 확인합니다. 비교 결과가 그대로 채점이 됩니다.

````python
answer = input("7 × 8 = ? ")
answer = int(answer)
print(answer == 56)   # 맞으면 True
````

## 2. 조건문: `if` 와 `if`-`else`

`if 조건:` 다음 줄을 **들여쓰기(4칸)**하여, 조건이 참일 때 실행할 문장을 씁니다. `else`는 거짓일 때 실행됩니다.

````python
age = 20
if age >= 19:
    print("성인입니다.")
print("프로그램 끝")
````

나머지 연산 `%`로 짝수·홀수를 판별합니다.

````python
number = 7
if number % 2 == 0:
    print("짝수입니다.")
else:
    print("홀수입니다.")
````

## 3. 논리 연산자

`and`(둘 다 참), `or`(하나라도 참), `not`(뒤집기)으로 여러 조건을 조합합니다.

````python
print(True and False)   # False
print(True or False)    # True
print(not True)         # False
````

**놀이공원 탑승** — 키가 120cm 이상이고 200cm 이하여야 탈 수 있습니다. `and`로 묶습니다.

````python
height = input("키(cm): ")
height = int(height)
if height >= 120 and height <= 200:
    print("탑승할 수 있습니다.")
else:
    print("탑승할 수 없습니다.")
````

## 4. `if`-`elif`-`else`

조건이 셋 이상이면 `elif`로 갈래를 늘립니다. 위에서부터 검사하여 **처음 참인 곳만** 실행합니다.

````python
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"
print(f"학점은 {grade}입니다.")
````

**BMI 판정** — 작은 기준부터 차례로 검사합니다.

````python
bmi = input("BMI: ")
bmi = float(bmi)
if bmi < 18.5:
    print("저체중")
elif bmi < 25:
    print("정상")
else:
    print("과체중")
````

## 5. 종합 활용

배운 것을 엮어 실전 프로그램을 만듭니다.

**사칙연산 계산기** — 두 수와 연산 기호를 입력받아 기호에 따라 다른 계산을 합니다.

````python
a = input("첫째 수: ")
a = int(a)
op = input("연산 기호(+, -, *, /): ")
b = input("둘째 수: ")
b = int(b)
if op == "+":
    print(a + b)
elif op == "-":
    print(a - b)
elif op == "*":
    print(a * b)
elif op == "/":
    print(a / b)
else:
    print("알 수 없는 연산입니다.")
````

**동전 던지기 게임** — `random` 모듈로 0/1을 무작위로 뽑습니다. `random`은 기본 내장 모듈이라 `import`만 하면 됩니다.

````python
import random
coin = random.randint(0, 1)   # 0 또는 1
if coin == 0:
    print("앞면입니다.")
else:
    print("뒷면입니다.")
````

**중첩 조건문 — 로그인** — 아이디가 맞은 다음에 비밀번호를 확인합니다.

````python
user_id = input("아이디: ")
if user_id == "kgu":
    password = input("비밀번호: ")
    if password == "1234":
        print("로그인 성공")
    else:
        print("비밀번호가 틀렸습니다.")
else:
    print("없는 아이디입니다.")
````
