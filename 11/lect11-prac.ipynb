## 1. `while`문의 구조와 3요소

`while` 문은 **조건이 참인 동안** 블록을 반복합니다. 실행 순서는 항상 같습니다.

1. 조건식을 검사한다.
2. 참이면 블록을 끝까지 실행하고 **1번으로 돌아간다.**
3. 거짓이면 블록을 건너뛰고 다음 코드로 넘어간다.

반복을 만들려면 **① 초기화 · ② 조건 검사 · ③ 변경** 세 가지가 모두 있어야 합니다.

````python
day = 1                          # ① 초기화
while day <= 7:                  # ② 조건 검사
    print(f"{day}일차 운동 완료")
    day += 1                     # ③ 변경
print("일주일 운동 끝!")
````

## 3. `for`문을 `while`문으로 바꾸기

`for` 문에서 `range()`가 해 주던 일(시작값·끝값 검사·다음 값)을 `while` 문에서는 직접 써야 합니다.

- 반복 **횟수**를 미리 안다 → `for` 문이 간결하다.
- 반복을 **멈출 조건**만 안다 → `while` 문이 알맞다.

````python
# for문
for i in range(1, 6):
    print(i)
````

````python
# 같은 일을 while문으로
i = 1             # 시작값
while i <= 5:     # 끝값 검사
    print(i)
    i += 1        # 다음 값으로
````

## 4. 무한 루프 피하기

조건이 **영원히 참**이면 루프가 끝나지 않습니다. 흔한 원인은 세 가지입니다. (아래 코드는 실행하지 말고 눈으로만 보세요.)

```python
# ① 변경을 빠뜨렸다
i = 1
while i <= 5:
    print(i)          # i += 1 이 없다

# ② 반대 방향으로 변경했다
i = 1
while i <= 5:
    print(i)
    i -= 1            # i가 점점 작아진다

# ③ 조건이 정확히 맞아떨어지지 않는다
i = 1
while i != 10:        # 1, 3, 5, 7, 9, 11, ... 10이 되는 순간이 없다
    print(i)
    i += 2
```

> ⚠️ **무한 루프에 빠졌다면** 셀 왼쪽의 **중지(■)** 버튼을 누르세요. 화면이 멈추면 **런타임 → 세션 다시 시작**을 누릅니다.

③은 `!=` 대신 **범위로 검사**(`<`)하면 해결됩니다.

````python
i = 1
while i < 10:         # "같지 않은 동안" 대신 "작은 동안"
    print(i)
    i += 2
````

## 6. 횟수를 모르는 반복

몇 번 반복할지 모르고 **멈출 조건만 알 때** `while` 문이 빛을 발합니다.

- **목표에 도달할 때까지**: 100만 원을 연 5%로 맡기면 몇 년 뒤 200만 원 이상이 될까?
- **숫자를 한 자리씩 쪼개기**: `n % 10`은 마지막 자리, `n // 10`은 마지막 자리를 떼어 낸 수
- **절반씩 줄이기**: 공이 튈 때마다 높이가 절반이 된다

````python
money = 1000000
year = 0
while money < 2000000:
    money = money * 105 // 100    # 1년치 이자(5%)를 붙인다
    year += 1
print(f"{year}년 뒤 {money}원")
````

````python
n = 2026
total = 0
while n > 0:
    digit = n % 10      # 마지막 자리
    total += digit
    n = n // 10         # 마지막 자리 떼어 내기
print(f"각 자리 숫자의 합: {total}")
````

````python
height = 100
bounce = 0
while height >= 1:
    height = height / 2
    bounce += 1
print(f"{bounce}번 튄 뒤 높이: {height}m")
````

## 7. 입력과 함께 쓰는 `while`

> 이 절의 코드 셀은 실행하면 **입력창**이 뜹니다. 안내에 따라 값을 입력해 보세요.

**종료값으로 입력 끝내기**: `0`을 입력할 때까지 지출 금액을 받아 합계를 구합니다. `while True:` + `break`로 쓰면 입력 코드를 한 번만 써도 됩니다.

````python
total = 0
while True:
    amount = input("지출 금액 (종료: 0): ")
    amount = int(amount)
    if amount == 0:
        break
    total += amount
print(f"총 지출: {total}원")
````

**올바른 값이 들어올 때까지 다시 묻기(입력 검증)**: 잘못된 값인 동안 계속 다시 묻습니다.

````python
score = input("만족도를 입력하시오 (1~5): ")
score = int(score)

while score < 1 or score > 5:
    print("1부터 5 사이의 수를 입력하세요.")
    score = input("만족도를 입력하시오 (1~5): ")
    score = int(score)

print(f"만족도 {score}점이 기록되었습니다.")
````

**메뉴를 반복해서 보여 주기**: 메뉴 번호는 `int()`로 바꾸지 않고 **문자열로 비교**하면 글자를 잘못 입력해도 오류가 나지 않습니다.

````python
balance = 0

while True:
    print("1. 입금  2. 출금  3. 잔액 조회  4. 종료")
    choice = input("번호를 선택하시오: ")

    if choice == "1":
        amount = input("입금액: ")
        amount = int(amount)
        balance += amount
    elif choice == "2":
        amount = input("출금액: ")
        amount = int(amount)
        if amount > balance:
            print("잔액이 부족합니다.")
        else:
            balance -= amount
    elif choice == "3":
        print(f"현재 잔액: {balance}원")
    elif choice == "4":
        print("이용해 주셔서 감사합니다.")
        break
    else:
        print("잘못된 번호입니다.")
````

## 8. `break`: 찾으면 바로 멈추기

````python
profits = [120, 80, -30, 50, -10]
i = 0
while i < len(profits):
    if profits[i] < 0:
        print(f"{i + 1}월에 처음 적자가 났다: {profits[i]}")
        break
    i += 1
````

## 9. `while`에서 `continue`를 쓸 때 주의할 점

`for` 문에서는 `continue`를 써도 다음 값이 자동으로 들어오지만, `while` 문에서는 **변경 코드를 건너뛰면 무한 루프**가 됩니다. (아래 코드는 실행하지 마세요.)

```python
i = 0
while i < 10:
    if i % 3 == 0:
        continue    # 아래의 i += 1 을 건너뛴다 → i는 0에 머문다 → 무한 루프!
    print(i)
    i += 1
```

> ⚠️ **규칙**: `while` 문에서 `continue`를 쓸 때는 조건 변수를 바꾸는 코드가 **`continue`보다 앞에** 있어야 합니다.

````python
i = 0
while i < 10:
    i += 1          # 먼저 변경한다
    if i % 3 == 0:
        continue    # 3의 배수는 출력하지 않고 다음 반복으로
    print(i)
````

## 10. 리스트가 빌 때까지 처리하기

`리스트.pop(0)`은 리스트의 **맨 앞 항목을 꺼내 돌려주고**, 리스트에서는 지웁니다. 반복할 때마다 리스트가 하나씩 줄어드므로 이것이 **③ 변경** 역할을 합니다.

````python
queue = ["민준", "서윤", "지호", "아린"]
while len(queue) > 0:
    person = queue.pop(0)     # 맨 앞 사람을 꺼낸다
    print(f"{person}님, 들어오세요. (남은 대기 {len(queue)}명)")
print("대기 인원이 없습니다.")
````

## 11. 종합 예제

> 아래 두 셀은 실행하면 **입력창**이 뜹니다.

**음료 자판기**: 가격표(딕셔너리)와 `while True:`·`break`·`continue`를 함께 씁니다. `while True:`에는 바꿀 조건 변수가 없으므로 `continue`를 써도 무한 루프 걱정이 없습니다.

````python
prices = {"콜라": 1500, "사이다": 1400, "생수": 900}
money = 5000
cart = []

while True:
    item = input("음료 이름 (종료: 끝): ")
    if item == "끝":
        break
    if item not in prices:
        print("없는 음료입니다.")
        continue
    if prices[item] > money:
        print(f"잔액이 부족합니다. (잔액 {money}원)")
        continue
    money -= prices[item]
    cart.append(item)
    print(f"{item} 구매! 잔액 {money}원")

print(f"구매한 음료: {cart}")
print(f"거스름돈: {money}원")
````

**기회가 정해진 숫자 맞히기**: `success`처럼 참·거짓 상태를 기억하는 변수를 두면, 루프가 끝난 뒤 **왜 끝났는지** 구분할 수 있습니다.

````python
import random

answer = random.randint(1, 50)
chance = 5
success = False      # 맞혔는지 기억하는 변수

while chance > 0 and not success:
    guess = input(f"1~50 사이의 수 (남은 기회 {chance}번): ")
    guess = int(guess)
    chance -= 1

    if guess == answer:
        success = True
    elif guess < answer:
        print("더 큰 수입니다.")
    else:
        print("더 작은 수입니다.")

if success:
    print("정답입니다!")
else:
    print(f"기회를 모두 썼습니다. 정답은 {answer}였습니다.")
````

## 정리

- **`while` 문의 3요소**: ① 초기화 · ② 조건 검사 · ③ 변경. 셋 중 하나라도 빠지면 무한 루프가 되기 쉽습니다.
- **무한 루프의 흔한 원인**: 변경 누락, 반대 방향 변경, 맞아떨어지지 않는 `!=` 조건.
- **횟수를 모르는 반복**: 목표 도달, 자릿수 쪼개기(`% 10`, `// 10`), 절반씩 줄이기.
- **입력과 함께**: 종료값으로 끝내기, 입력 검증, `while True:` 메뉴 반복.
- **`break`** 는 루프를 완전히 끝내고, **`continue`** 는 이번 반복만 건너뜁니다. `while` 문에서는 변경 코드를 `continue`보다 앞에 둡니다.
