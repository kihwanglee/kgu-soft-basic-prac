### 문제 3. 리스트에 번호 붙여 출력 (for·인덱스)  〔🔲 빈칸 채우기〕

메뉴를 `1번`, `2번`, `3번` 처럼 **순서를 붙여** 출력하려고 한다. 인덱스는 0부터 시작하므로 빈칸을 채워 순서가 1부터 나오게 하시오.

**예시 결과:**
```
1번: 아메리카노
2번: 라떼
3번: 녹차
```

````python
menu = ["아메리카노", "라떼", "녹차"]
for i in range(len(menu)):
    print(f"{____}번: {menu[i]}")
````
