## 1. 리스트 만들기

**리스트**는 여러 값을 하나의 이름으로 묶는 자료형입니다. 대괄호 `[]` 안에 값을 쉼표로 나열합니다.

````python
scores = [90, 85, 70, 95]
fruits = ["사과", "바나나", "포도"]
print(scores)
print(fruits)
````

### 인덱싱 — 자리 번호로 값 꺼내기

각 값에는 **인덱스**(자리 번호)가 있고 **0부터** 시작합니다. 음수 인덱스는 뒤에서부터 셉니다(`-1`이 마지막).

````python
scores = [90, 85, 70, 95]
print(scores[0])    # 90  (첫 번째)
print(scores[2])    # 70  (세 번째)
print(scores[-1])   # 95  (마지막)
````

### 슬라이싱 — 일부분 잘라내기

`리스트[시작:끝]` 로 일부를 잘라냅니다. **시작은 포함, 끝은 제외**합니다.

````python
scores = [90, 85, 70, 95]
print(scores[1:3])   # [85, 70]
print(scores[:2])    # [90, 85]
print(scores[::-1])  # [95, 70, 85, 90]  (역순)
````

### 수정·길이·메서드

인덱스로 값을 바꾸고, `len()`으로 개수를, 메서드로 추가·삭제·정렬을 합니다.

````python
scores = [90, 85, 70, 95]
scores[2] = 100          # 세 번째 값 수정
print(scores)            # [90, 85, 100, 95]
print(len(scores))       # 4

fruits = ["사과", "바나나"]
fruits.append("포도")     # 맨 뒤에 추가
fruits.remove("바나나")   # 값으로 삭제
print(fruits)            # ['사과', '포도']
print("사과" in fruits)  # True
````

## 2. `for` 반복문

`for 변수 in 리스트:` 는 값을 하나씩 꺼내 반복합니다. 조건문처럼 콜론 `:` 과 들여쓰기를 씁니다.

````python
scores = [90, 85, 70, 95]
for score in scores:
    print(score)
````

````python
fruits = ["사과", "바나나", "포도"]
for fruit in fruits:
    print(f"나는 {fruit}를 좋아합니다.")
````

### `range()` — 정해진 횟수 반복

`range(n)`은 `0`~`n-1`, `range(시작, 끝)`은 시작~끝 직전, `range(시작, 끝, 간격)`으로 건너뛰거나 거꾸로 셉니다.

````python
for i in range(5):
    print(i)        # 0 1 2 3 4
````

````python
for i in range(1, 6):
    print(i)        # 1 2 3 4 5

for i in range(0, 11, 2):
    print(i)        # 0 2 4 6 8 10
````

## 3. 리스트와 `for` 문 활용

리스트의 값을 한꺼번에 처리합니다.

**합계와 평균** — 합을 담을 변수를 `0`에서 시작해 하나씩 더합니다.

````python
scores = [90, 85, 70, 95]
total = 0
for score in scores:
    total += score
print(f"합계: {total}, 평균: {total / len(scores):.1f}")
# sum(), max(), min() 함수로도 구할 수 있습니다.
print(sum(scores), max(scores), min(scores))
````

**조건에 맞는 값만** — `for` 안에 `if`를 써서 60점 이상만 셉니다.

````python
scores = [90, 55, 70, 40, 85]
count = 0
for score in scores:
    if score >= 60:
        count += 1
print(f"60점 이상: {count}명")
````

**조건에 맞는 값 모으기** — `for` + `if` + `append` 로 새 리스트를 만듭니다.

````python
evens = []
for i in range(1, 11):
    if i % 2 == 0:
        evens.append(i)
print(evens)    # [2, 4, 6, 8, 10]
````
